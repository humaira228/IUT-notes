## Cell 1: Import Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import wavfile
```

## Task 1: Spectral Analysis

### Cell 2: Read Audio

```python
# Read audio
fs, data = wavfile.read("Lab 4.wav")

print("Sampling Frequency:", fs)
print("Number of Samples:", len(data))
```

### Cell 3: Convert Stereo to Mono (if needed)

```python
if len(data.shape) > 1:
    data = data.mean(axis=1)

data = data.astype(float)
```

### Cell 4: Plot the Signal

```python
time = np.arange(len(data))/fs

plt.figure(figsize=(12,4))
plt.plot(time, data)
plt.title("Audio Signal")
plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()
```

### Cell 5: Spectral Analysis using rFFT

```python
fft_values = np.fft.rfft(data)

frequencies = np.fft.rfftfreq(len(data), d=1/fs)

magnitude = np.abs(fft_values)
```

### Cell 6: Plot Frequency Spectrum

```python
plt.figure(figsize=(12,5))
plt.plot(frequencies, magnitude)

plt.title("Magnitude Spectrum")
plt.xlabel("Frequency (Hz)")
plt.ylabel("Magnitude")
plt.grid(True)
plt.show()
```

### Cell 7: Find Tuning Fork Frequency

```python
peak_index = np.argmax(magnitude[1:]) + 1

peak_frequency = frequencies[peak_index]

print("Estimated Tuning Fork Frequency =", peak_frequency, "Hz")
```

## Task 2

### Cell 8: Forward DFT From Scratch

```python
def myDFT(x):

    x = np.asarray(x, dtype=float)

    N = len(x)

    ReX = np.zeros(N//2 + 1)

    ImX = np.zeros(N//2 + 1)

    for k in range(N//2 + 1):

        for i in range(N):

            angle = 2*np.pi*k*i/N

            ReX[k] += x[i]*np.cos(angle)

            ImX[k] -= x[i]*np.sin(angle)

    return ReX, ImX
```

### Cell 9: Inverse DFT From Scratch

```python
def myIDFT(ReX, ImX):

    M = len(ReX)

    N = (M-1)*2

    Re = ReX.copy()/(N/2)

    Im = -ImX.copy()/(N/2)

    Re[0] /= 2

    Re[-1] /= 2

    x = np.zeros(N)

    for i in range(N):

        for k in range(M):

            angle = 2*np.pi*k*i/N

            x[i] += Re[k]*np.cos(angle)

            x[i] += Im[k]*np.sin(angle)

    return x
```

### Cell 10: Verification (Figure 8-6)

```python
N = 32

x = np.zeros(N)

x[0] = 32
```

### Cell 11: Apply DFT

```python
ReX, ImX = myDFT(x)

print("Real Part")
print(ReX)

print()

print("Imaginary Part")
print(ImX)
```

### Cell 12: Apply IDFT

```python
reconstructed = myIDFT(ReX, ImX)
```
